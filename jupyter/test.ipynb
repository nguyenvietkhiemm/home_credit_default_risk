```python
def read(filename):
    try:
        with open(filename, 'r') as f:
            features = [line.strip() for line in f]
        return features
    except FileNotFoundError:
        print(f"Lỗi: Không tìm thấy file {filename}")
        return None
    except Exception as e:
        print(f"Lỗi khi đọc file {filename}: {e}")
        return None
```

```python
from rapidfuzz import fuzz

def find_fuzzy_matches(list1, list2, threshold=96):
    matches = []
    for a in list1:
        for b in list2:
            score = fuzz.ratio(a, b)
            if threshold <= score <= 100:
                matches.append((a, b, score))
    # Sắp xếp theo score giảm dần
    matches.sort(key=lambda x: x[2], reverse=True)
    return matches

def find_common_elements(list1, list2):
    return list(set(list1) & set(list2))

lb = pd.read_csv(ROOT + "/data/lb804.csv")["feature"].str.replace(r'[+()\-,. ]', '_', regex=True).replace(r'[_m_]', '_s_').to_list()
li = pd.Series(lgbm_imp_sorted_2["feature"].unique()).str.replace(r'[+()\-,. ]', '_', regex=True).tolist()

res=find_fuzzy_matches(lb, li)
print(*res, len(res),sep="\n")
```

```python
def filter_by_threshold(chunk_data, per=80):
    a = chunk_data.shape
    threshold_gain = min(np.percentile(chunk_data["importance_gain"], per), 1000)
    threshold_auc_valid = min(np.percentile(chunk_data["auc_valid"], per), 0.6)
    threshold_split = min(np.percentile(chunk_data["importance_split"], per), 2)
    res = chunk_data[(chunk_data["importance_gain"] > threshold_gain) & (chunk_data["auc_valid"] > threshold_auc_valid) & (chunk_data["importance_split"] > threshold_split)]
    b = res.shape
    print(f"{a} => {b}")
    return res # lấy threshold là phân vị default 80%
```

# lấy phân vị cao hơn cho low importance
