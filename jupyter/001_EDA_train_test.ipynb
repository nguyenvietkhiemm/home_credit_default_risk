## categorical features

```python
import lightgbm as lgb
import multiprocessing
from gplearn.genetic import SymbolicTransformer
```

```python
SEED = 71
param = {
    'objective': 'binary', 'metric': 'auc', 'learning_rate': 0.05, 'max_depth': -1, 'num_leaves': 255, 'max_bin': 255, 'colsample_bytree': 0.5, 'subsample': 0.5, 'nthread':
    multiprocessing.cpu_count(), 'bagging_freq': 1, 'seed': SEED
}

categorical_feature = [
    'NAME_CONTRACT_TYPE', 'CODE_GENDER', 'FLAG_OWN_CAR', 'FLAG_OWN_REALTY', 'NAME_TYPE_SUITE', 'NAME_INCOME_TYPE', 'NAME_EDUCATION_TYPE', 'NAME_FAMILY_STATUS', 'NAME_HOUSING_TYPE', 'OCCUPATION_TYPE',
    'WEEKDAY_APPR_PROCESS_START', 'ORGANIZATION_TYPE', 'FONDKAPREMONT_MODE', 'HOUSETYPE_MODE', 'WALLSMATERIAL_MODE', 'EMERGENCYSTATE_MODE'
]
```

```python
X = train.drop(["TARGET", "data", "SK_ID_CURR"], axis=1)
y = train["TARGET"]
```

```python
for col in X.select_dtypes(include=['object']).columns:
    X[col] = X[col].astype('category')
```

### cv (cross validation) benchmark

```python
dtrain = lgb.Dataset(X, y, categorical_feature=categorical_feature)
```

```python
ret = lgb.cv(param, dtrain, 9999, nfold=5, callbacks=[lgb.early_stopping(50), lgb.log_evaluation(10)], seed=SEED)
```

```python
ret['valid auc-mean'][-1]
```

### cv encoding

#### leave one out encode

```python
import category_encoders as ce
```

```python
enc = ce.LeaveOneOutEncoder(cols=categorical_feature)  # looe
```

```python
X = enc.fit_transform(X, y)
```

```python
dtrain = lgb.Dataset(X, y, categorical_feature=categorical_feature)
```

```python
ret = lgb.cv(param, dtrain, 9999, nfold=5, callbacks=[lgb.early_stopping(50), lgb.log_evaluation(10)], seed=SEED)
```

```python
ret["valid auc-mean"][-1]
```

#### label encode

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()  # label encode
```

```python
_X = X.copy()
_y = y.copy()
```

```python
for c in categorical_feature:
    most_frequent = _X[c].mode()[0]
    _X[c].fillna(most_frequent, inplace=True)
    _X[c] = le.fit_transform(_X[c])
```

```python
_X.head()
```

```python
dtrain = lgb.Dataset(_X, _y, categorical_feature=categorical_feature)
ret = lgb.cv(param, dtrain, 9999, nfold=5, callbacks=[lgb.early_stopping(50), lgb.log_evaluation(10)], seed=SEED)
```

```python
ret["valid auc-mean"][-1]
```

#### target encode

```python
enc = ce.TargetEncoder(cols=categorical_feature)
```

```python
_X = X.copy()
_y = y.copy()
```

```python
_X = enc.fit_transform(_X, _y)
```

```python
for c in categorical_feature:
    most_frequent = _X[c].mode()[0]
    _X[c].fillna(most_frequent, inplace=True)
    _X[c] = le.fit_transform(_X[c])
```

```python
dtrain = lgb.Dataset(X, y, categorical_feature=categorical_feature)
```

```python
ret = lgb.cv(param, dtrain, 9999, nfold=5, callbacks=[lgb.early_stopping(50), lgb.log_evaluation(10)], seed=SEED)
```

```python
ret["valid auc-mean"][-1]
```
